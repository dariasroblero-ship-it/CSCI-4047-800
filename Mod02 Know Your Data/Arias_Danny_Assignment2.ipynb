```python
import pandas as pd, numpy as np, matplotlib.pyplot as mpp

df = pd.read_csv("./Income Data.csv")
print(f"Extracted {len(df):,} rows and {len(df.columns)} columns")
df.head()
```

3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
min = np.min(df.income)
q1 = np.percentile(df.income, 25)
med = np.median(df.income)
mn = round(np.mean(df.income), 2)
q3 = np.percentile(df.income, 75)
max = np.max(df.income)

print('------------ Income -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mn}")
print(f"Q3: {q3}")
print(f"Max: {max}")


min = np.min(df.education)
q1 = np.percentile(df.education, 25)
med = np.median(df.education)
mn = round(np.mean(df.education), 2)
q3 = np.percentile(df.education, 75)
max = np.max(df.education)

print('------------ Education -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mn}")
print(f"Q3: {q3}")
print(f"Max: {max}")

min = np.min(df.prestige)
q1 = np.percentile(df.prestige, 25)
med = np.median(df.prestige)
mn = round(np.mean(df.prestige), 2)
q3 = np.percentile(df.prestige, 75)
max = np.max(df.prestige)

print('------------ Prestige -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Mean: {mn}")
print(f"Q3: {q3}")
print(f"Max: {max}")
```

4. Plot a *histogram* for **prestige**
5. Plot a *histogram* for **education**

```python
print("---------------------------------------Prestige-------------------------------")
mpp.hist(df['prestige'], bins=7, color='orange', alpha=0.5)
mpp.show()

print("---------------------------------------Prestige-------------------------------")
mpp.hist(df['education'], bins=7, color='orange', alpha=0.5)
mpp.show()
```

6. Plot a *scatter plot* for **income** and **education**
7. Plot a *scatter plot* for **education** and **prestige**

```python
print("---------------------------------------Income and education-------------------------------")
mpp.scatter(x=df['education'], y=df['income'], alpha=0.3)
mpp.title("Education compared to income")
mpp.xlabel('education')
mpp.ylabel("income")
mpp.show()

print("---------------------------------------Education and Prestige-------------------------------")
mpp.scatter(x=df['prestige'], y=df['education'], alpha=0.3)
mpp.title("Prestige compared to education")
mpp.xlabel('prestige')
mpp.ylabel("education")
mpp.show()
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
standard_dev = df['income'].std()
print()
print(f"Standard Deviation:  {standard_dev:.2f}" )

variance = standard_dev ** 2
print(f"Variance:            {variance:.2f}")
```
